# Example with the Kiutra L-Type Rapid

This notebook shows how to drive a Kiutra L-Type Rapid cryostat through the
`LTypeRapid` driver, and how to reach the underlying `kiutra_api` controllers
directly when the driver does not expose something.

It needs a live system: set `KIUTRA_ADDRESS` below to the address of yours.
The saved output was recorded on a real cryostat, and the notebook is not
re-executed when the documentation is built.


## Import

```python
import qcodes as qc
from qcodes_contrib_drivers.drivers.Kiutra.L_Type_Rapid import LTypeRapid
# Address of the cryostat this notebook talks to.
KIUTRA_ADDRESS = "192.168.11.20"
KIUTRA_PORT = 1006
```

```python
try:
    kiutra = LTypeRapid("kiutra", KIUTRA_ADDRESS)
except KeyError as e:
    print(f"{e}")
temp = kiutra.temperature_control.controller
```

```python
setpoint = 1
ramp = 0.01
mode = "ramp"
print(
    kiutra.temperature_control.controller.call_method(
        "propose_control_mode", setpoint=setpoint, ramp=ramp, mode=mode
    )
)
print(
    kiutra.temperature_control.controller.call_method(
        "propose_control_settings", setpoint=setpoint, ramp=ramp, mode=mode
    )
)
```

```python
print(temp.call_method("start_control", setpoint=2.91, ramp=0.02, control_mode="123"))
```

```python
temp.query_value('status')
```

```python
import pandas as pd


pd.DataFrame(kiutra.temperature_control.controller.api).transpose()
```

```python
from kiutra_api.controller_interfaces import TemperatureControl
import pandas as pd
tc = TemperatureControl("temperature_control", KIUTRA_ADDRESS, KIUTRA_PORT)
# Display the API as a DataFrame for better readability
pd.DataFrame(tc.api).transpose()
```

```python
tc.query_value('guard_min_temp_cadr')
```

```python
from kiutra_api.controller_interfaces import MagnetControl
import pandas as pd
mc = MagnetControl("sample_magnet", KIUTRA_ADDRESS, KIUTRA_PORT)
pd.DataFrame(mc.api).transpose()
```
