# QCoDeS example with the QuTech IVVI module catalogue

This notebook lists the QuTech IVVI rack modules that the package knows about, and shows how to build one of them with its QCoDeS driver.

## The catalogue

`CATALOGUE` has one entry per module type. Each entry gives the module kind, the slot kind it fits, its front-panel ports and its knobs. A knob lists the values it accepts and its starting value. `catalogue_json()` returns the same data as plain dictionaries, for tools that cannot import the classes.

## Virtual drivers

The module drivers are virtual. They record the slot and the knob positions you set, and they do not talk to hardware. Set the physical module by hand to match the driver. The driver cannot check the module.

### Table of module types

Each line shows a module type, its kind, its slot kind, its port ids and its knobs. A knob appears as its name followed by the values it accepts, for example `output_range[30, 60, 90]`.

```python
from qcodes_contrib_drivers.drivers.QuTech.catalogue import CATALOGUE

print("type | label | kind | slot | ports | knobs")
print("-" * 80)
for spec in CATALOGUE.values():
    ports = ", ".join(port.id for port in spec.ports) or "none"
    knobs = " ".join(f"{knob.name}[{', '.join(str(v) for v in knob.values)}]" for knob in spec.knobs) or "none"
    print(f"{spec.type} | {spec.label} | {spec.kind} | {spec.slot_kind} | {ports} | {knobs}")
```

### Build an S1h

The S1h is the high-voltage backgate source. The next cell creates one named `s1h_demo`, sets its output range to 60 V, prints its settings and closes it.

```python
from qcodes_contrib_drivers.drivers.QuTech.S1h import S1h

s1h = S1h("s1h_demo")
s1h.output_range("60")
s1h.print_readable_snapshot()
s1h.close()
```
